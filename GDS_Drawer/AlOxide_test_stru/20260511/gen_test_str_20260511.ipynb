```python
import gdstk

def create_gds(SiO2_OFFSET=10):
    # Create a new GDSII library
    lib = gdstk.Library()

    # Create a new cell for the test structure
    cell = gdstk.Cell('TEST_STRUCTURE')

    # Define the layers based on your provided panel
    LAYER_MA_GREY = 10
    LAYER_MA_ORANGE = 11
    LAYER_EBL_GREY = 12
    LAYER_EBL_ORANGE = 13
    LAYER_SIO2 = 20  # New layer for SiO2

    # Calculate the inset for each side to achieve: Oxide size = pad size - SiO2_OFFSET
    inset = SiO2_OFFSET / 2.0

    # ==========================================
    # 1. MA_GREY (Layer 10) - Horizontal (Left)
    # ==========================================
    # Narrow arm (Width 5, Length 20, Gap from origin is 2.5)
    ma_grey_arm1 = gdstk.rectangle((-22.5, -2.5), (-5, 2.5), layer=LAYER_MA_GREY)
    
    # Wide arm (Width 20, Length 100)
    ma_grey_arm2 = gdstk.rectangle((-122.5, -10.0), (-22.5, 10.0), layer=LAYER_MA_GREY)
    
    # Large Pad (160x160)
    ma_grey_pad = gdstk.rectangle((-282.5, -80.0), (-122.5, 80.0), layer=LAYER_MA_GREY)

    # SiO2 under MA_GREY Pad
    sio2_grey_pad = gdstk.rectangle(
        (-282.5 + inset, -80.0 + inset), 
        (-122.5 - inset, 80.0 - inset), 
        layer=LAYER_SIO2
    )

    # ==========================================
    # 2. MA_ORANGE (Layer 11) - Vertical (Bottom)
    # ==========================================
    # Narrow arm (Width 5, Length 20, Gap from origin is 2.5)
    ma_orange_arm1 = gdstk.rectangle((-2.5, -22.5), (2.5, -5), layer=LAYER_MA_ORANGE)
    
    # Wide arm (Width 20, Length 100)
    ma_orange_arm2 = gdstk.rectangle((-10.0, -122.5), (10.0, -22.5), layer=LAYER_MA_ORANGE)
    
    # Large Pad (160x160)
    ma_orange_pad = gdstk.rectangle((-80.0, -282.5), (80.0, -122.5), layer=LAYER_MA_ORANGE)

    # SiO2 under MA_ORANGE Pad
    sio2_orange_pad = gdstk.rectangle(
        (-80.0 + inset, -282.5 + inset), 
        (80.0 - inset, -122.5 - inset), 
        layer=LAYER_SIO2
    )

    # ==========================================
    # 3. EBL_GREY (Layer 12) - Horizontal Line
    # ==========================================
    # Center Y=0. Width = 0.05. Length = 45 (Spans symmetrically from inside the MA arm to past the center)
    ebl_grey_line = gdstk.rectangle((-22.5, -0.025), (22.5, 0.025), layer=LAYER_EBL_GREY)

    # ==========================================
    # 4. EBL_ORANGE (Layer 13) - Vertical Line
    # ==========================================
    # Center X=0. Width = 0.05. Length = 45 
    ebl_orange_line = gdstk.rectangle((-0.025, -22.5), (0.025, 22.5), layer=LAYER_EBL_ORANGE)

    # Add all polygons to the cell
    cell.add(
        ma_grey_arm1, ma_grey_arm2, ma_grey_pad, sio2_grey_pad,
        ma_orange_arm1, ma_orange_arm2, ma_orange_pad, sio2_orange_pad,
        ebl_grey_line, ebl_orange_line
    )

    # Add the cell to the library
    lib.add(cell)

    # Save the layout to a file
    filename = 'test_structure_gdstk.gds'
    lib.write_gds(filename)
    print(f"GDS plot successfully generated and saved as '{filename}'")

if __name__ == "__main__":
    # You can easily change the SiO2 offset here if needed
    create_gds(SiO2_OFFSET=10)
```

```python
import gdstk

def create_gds(SiO2_OFFSET=10):
    # Create a new GDSII library
    lib = gdstk.Library()

    # Create a new cell for a single test structure
    single_cell = gdstk.Cell('SINGLE_TEST_STRUCTURE')

    # Define the layers based on your provided panel
    LAYER_MA_GREY = 10
    LAYER_MA_ORANGE = 11
    LAYER_EBL_GREY = 30
    LAYER_EBL_ORANGE = 31
    LAYER_SIO2 = 20  # Layer for SiO2

    # Calculate the inset for each side to achieve: Oxide size = pad size - SiO2_OFFSET
    inset = SiO2_OFFSET / 2.0

    # ==========================================
    # 1. MA_GREY (Layer 10) - Horizontal (Left)
    # ==========================================
    ma_grey_arm1 = gdstk.rectangle((-22.5, -2.5), (-5, 2.5), layer=LAYER_MA_GREY)
    ma_grey_arm2 = gdstk.rectangle((-122.5, -10.0), (-22.5, 10.0), layer=LAYER_MA_GREY)
    ma_grey_pad = gdstk.rectangle((-282.5, -80.0), (-122.5, 80.0), layer=LAYER_MA_GREY)

    sio2_grey_pad = gdstk.rectangle(
        (-282.5 + inset, -80.0 + inset), 
        (-122.5 - inset, 80.0 - inset), 
        layer=LAYER_SIO2
    )

    # ==========================================
    # 2. MA_ORANGE (Layer 11) - Vertical (Bottom)
    # ==========================================
    ma_orange_arm1 = gdstk.rectangle((-2.5, -22.5), (2.5, -5), layer=LAYER_MA_ORANGE)
    ma_orange_arm2 = gdstk.rectangle((-10.0, -122.5), (10.0, -22.5), layer=LAYER_MA_ORANGE)
    ma_orange_pad = gdstk.rectangle((-80.0, -282.5), (80.0, -122.5), layer=LAYER_MA_ORANGE)

    sio2_orange_pad = gdstk.rectangle(
        (-80.0 + inset, -282.5 + inset), 
        (80.0 - inset, -122.5 - inset), 
        layer=LAYER_SIO2
    )

    # ==========================================
    # 3. EBL_GREY (Layer 12) - Horizontal Line
    # ==========================================
    ebl_grey_line = gdstk.rectangle((-22.5, -0.025), (22.5, 0.025), layer=LAYER_EBL_GREY)

    # ==========================================
    # 4. EBL_ORANGE (Layer 13) - Vertical Line
    # ==========================================
    ebl_orange_line = gdstk.rectangle((-0.025, -22.5), (0.025, 22.5), layer=LAYER_EBL_ORANGE)

    # Add all polygons to the single cell
    single_cell.add(
        ma_grey_arm1, ma_grey_arm2, ma_grey_pad, sio2_grey_pad,
        ma_orange_arm1, ma_orange_arm2, ma_orange_pad, sio2_orange_pad,
        ebl_grey_line, ebl_orange_line
    )

    # ==========================================
    # CREATE THE 2x2 ARRAY
    # ==========================================
    # Create a TOP cell that will contain the array
    main_cell = gdstk.Cell('MAIN_ARRAY')

    # Set array spacing (Distance between the (0,0) origins of each repeating unit)
    # 400um ensures the 362.5um-wide structures do not overlap
    spacing_x = 400
    spacing_y = 400

    # Create a reference to the single cell, arranged in a 2x2 grid
    array_ref = gdstk.Reference(
        single_cell, 
        origin=(0, 0), 
        columns=2, 
        rows=2, 
        spacing=(spacing_x, spacing_y)
    )

    # Add the array reference to the main cell
    main_cell.add(array_ref)

    # Add BOTH the single component cell and the main array cell to the library
    lib.add(single_cell, main_cell)

    # Save the layout to a file
    filename = 'test_structure_2x2_array.gds'
    lib.write_gds(filename)
    print(f"2x2 GDS plot successfully generated and saved as '{filename}'")

if __name__ == "__main__":
    create_gds(SiO2_OFFSET=10)
```
