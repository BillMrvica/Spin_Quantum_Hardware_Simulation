```python
import gdstk
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

# Load GDS
lib = gdstk.read_gds('WGL_G12_EBL_0729.GDS')
top = lib.top_level()[0]

fig, ax = plt.subplots(figsize=(10, 10))

# Plot polygons
for poly in top.polygons:
    ax.add_patch(Polygon(poly.points, closed=True, alpha=0.5, facecolor='blue', edgecolor='none'))

# Plot paths
for path in top.paths:
    for p in path.to_polygons():
        ax.add_patch(Polygon(p, closed=True, alpha=0.5, facecolor='red', edgecolor='none'))

# Add text labels
for label in top.labels:
    ax.text(label.origin[0], label.origin[1], label.text, fontsize=8, color='black', ha='center', va='center')

ax.set_xlim(0, 20000)
ax.set_ylim(0, 20000)
ax.set_aspect('equal')
plt.title('GDS Layout Overview')
plt.grid(True)
plt.show()
```
