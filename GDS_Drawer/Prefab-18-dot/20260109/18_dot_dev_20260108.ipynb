```python
import os
import sys
import gdstk
import numpy as np
import matplotlib.pyplot as plt

# Add current directory to path to import the generator
sys.path.append(os.getcwd())

from quantum_dot_pad_generator_18dot import QuantumDotPadGenerator18Dot

%matplotlib inline
```

```python
# 1. Initialize the generator
generator = QuantumDotPadGenerator18Dot()

# 2. Generate the 18-dot device layout
device_lib, device_cell, device_connection_points = generator.generate_quantum_dot_layout()

# Visualize the device core
generator._plot_gds(device_cell, title="18-Dot Device Core with Leads")
```

```python
# 3. Generate the pad frame layout
pad_lib, pad_cell, pad_connection_points, active_center, all_pads_info = generator.generate_pad_frame_layout(
    layout_width=2900, 
    layout_height=2900, 
    active_size=400
)

# Visualize and save the pad frame
generator.visualize_and_save_pad_frame(pad_lib, pad_cell, "18-Dot Pad Frame", "18dot_pad_frame.gds")
```

```python
# 4. Assemble the device and the pad frame
# Use the built-in assembly method for better reliability
assembly_lib, assembly_cell = generator.assemble_device_and_pad_frame()

if assembly_cell is not None:
    # 5. Visualize and save the final assembly
    # The new visualization method provides both full view and zoomed-in center view
    generator.visualize_and_save_assembly(assembly_lib, assembly_cell, "18-Dot Full Assembly", "18dot_full_assembly.gds")
    print("Successfully assembled and saved 18dot_full_assembly.gds")
else:
    print("Error: Assembly failed.")
```
